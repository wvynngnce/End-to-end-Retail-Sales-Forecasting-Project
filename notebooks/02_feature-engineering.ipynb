```python
#FEATURE ENGINEERING
1. Lag 1, 7, 14, dan 28
2. Rolling mean 7 dan 28
3. day of week, month and year
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
model_data = pd.read_csv(
    r"C:\Projek\Machine Learning\portfolio\01\retail-sales-forecasting\data\processed\daily_family_sales.csv",
    parse_dates=["date"]
)

model_data.head()
```

PENGURUTAN DATA

```python
model_data = model_data.sort_values(
    ["family", "date"]
).reset_index(drop=True)

print("Dataset shape:", model_data.shape)
print(model_data.head())
```

```python
print(model_data.tail())
```

FITUR KALENDER

```python
model_data['day_of_week'] = model_data['date'].dt.dayofweek
model_data['day_of_month'] = model_data['date'].dt.day
model_data['month'] = model_data['date'].dt.month
model_data['year'] = model_data['date'].dt.year
```

LAG FEATURES

```python
for lag in [1, 7, 14, 28]:
    model_data[f'lag_{lag}'] = (
        model_data.groupby('family')['sales'].shift(lag)
    )
```

ROLLING MEAN

```python
for window in [7, 28]:
    model_data[f"rolling_mean_{window}"] = (
        model_data.groupby('family')['sales']
        .transform(
            lambda x: x.shift(1).rolling(window=window).mean()
        )
    )
```

```python
print("Dataset shape: ", model_data.shape)

print("\nMissing values: ")
print(model_data.isnull().sum())

model_data.head(35)
```

```python
model_data = model_data.dropna().reset_index(drop=True)

print("Final dataset shape: ", model_data.shape)
print("Missing values: ", model_data.isnull().sum())
```

SAVE DATASET

```python
model_data.to_csv(
    r"C:\Projek\Machine Learning\portfolio\01\data\processed\forecasting_features.csv",
    index=False
)

print("Feature engineering completed. The dataset has been saved to 'forecasting_features.csv'.")
```
